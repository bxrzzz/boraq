```python
# Telco ma'lumotlar to'plamini internetdan yuklab olayapman
import pandas as pd
url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)
```

```python
# select_dtypes yordamida faqat matn (object) turidagi ustunlarni ajratib, ro'yxatini chiqarayapman
obj_cols = df.select_dtypes(include='object').columns.tolist()
print('Total object columns:', len(obj_cols))
print(obj_cols)
```

```python
# gender ustunida Male va Female nechtadan borligini sanayapman
df['gender'].value_counts()
```

```python
# InternetService ustunini pd.get_dummies orqali har bir qiymat uchun alohida 0/1 ustunga aylantirayapman
dummies = pd.get_dummies(df['InternetService'], prefix='Internet')
print('Shape:', dummies.shape)
dummies.head()
```

```python
# drop_first=True qo'shib, ortiqcha (birinchi) ustunni olib tashlab, dublikat ma'lumotni kamaytirayapman
dummies2 = pd.get_dummies(df['InternetService'], prefix='Internet', drop_first=True)
print('With drop_first:', dummies2.shape)
dummies2.head()
```

```python
# sklearn kutubxonasidagi OneHotEncoder yordamida TechSupport ustunini kodlashtirayapman
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output=False)
encoded = ohe.fit_transform(df[['TechSupport']])
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(['TechSupport']))
encoded_df.head()
```

```python
# Partner ustunidagi Yes/No qiymatlarini map yordamida 1/0 raqamlarga aylantirayapman
df['Partner_bin'] = df['Partner'].map({'No': 0, 'Yes': 1})
df[['Partner', 'Partner_bin']].head()
```

```python
# Bir nechta ustunni bir vaqtning o'zida get_dummies orqali kodlashtirib, natijadagi ustunlar sonini ko'rayapman
multi = pd.get_dummies(df[['gender', 'InternetService', 'TechSupport', 'Contract']], drop_first=True)
print('Shape:', multi.shape)
multi.head()
```
