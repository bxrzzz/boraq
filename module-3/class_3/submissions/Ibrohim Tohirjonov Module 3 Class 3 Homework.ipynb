```python
# Step 0 --- Load the data
# This code imports the Pandas library to handle structured data within Python.
# It then pulls the raw Telco Customer Churn dataset from a remote IBM repository and loads it into a DataFrame for analysis.

import pandas as pd
url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)
```

```python
# Step 1 --- List all object (text) columns
# This code identifies all columns in the DataFrame that contain categorical text or object data types and saves their names into a list.
# It then prints the total count of these non-numeric columns followed by their specific name labels for structural evaluation.

obj_cols = df.select_dtypes(include='object').columns.tolist()
print('Total object columns:', len(obj_cols))
print(obj_cols)
```

```python
# Step 2 --- value_counts on gender
# This code computes the total frequency distribution for each unique category present within the 'gender' column.
# It returns a breakdown of the specific row counts to help analyze the baseline gender balance of the dataset.

df['gender'].value_counts()
```

```python
# Step 3 --- One-hot encode InternetService with pd.get_dummies
# This code converts the categorical 'InternetService' column into binary dummy variables using one-hot encoding, adding a custom prefix to the new features.
# It then prints the structural shape of the resulting encoded DataFrame and displays a quick preview of its top rows to verify the transformation.

dummies = pd.get_dummies(df['InternetService'], prefix='Internet')
print('Shape:', dummies.shape)
dummies.head()
```

```python
# Step 4 --- Same column but with drop_first=True
# This code applies one-hot encoding to the 'InternetService' column while dropping the first category to prevent multi-collinearity issues.
# It then prints the newly reduced column dimensions and previews the top rows to confirm the creation of the final indicator variables.

dummies2 = pd.get_dummies(df['InternetService'], prefix='Internet', drop_first=True)
print('With drop_first:', dummies2.shape)
dummies2.head
```

```python
# Step 5 --- OneHotEncoder on techSupport
# This code applies a scikit-learn OneHotEncoder to convert the 'TechSupport' column into a dense array of binary indicator variables.
# It then wraps the matrix in a Pandas DataFrame using the dynamically generated category names as headers and previews the top rows to verify the encoding structure.

from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output=False)
encoded = ohe.fit_transform(df[['TechSupport']])
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(['TechSupport']))
encoded_df.head()
```

```python
# Step 6 --- Binary Yes/No mapping
# This code converts the categorical 'Partner' column into a binary numeric format by mapping 'No' to 0 and 'Yes' to 1.
# It then displays a quick preview of both the original and newly mapped columns side-by-side to verify the transformation.

df['Partner_bin'] = df['Partner'].map({'No': 0, 'Yes': 1})
df[['Partner', 'Partner_bin']].head()
```

```python
# Step 7 --- Encode many columns at once
# This code extracts multiple categorical columns from the DataFrame and applies one-hot encoding to them simultaneously.
# It automatically drops the first category of each feature to reduce redundancy, prints the resulting shape, and displays the top rows for verification.

multi = pd.get_dummies(df[['gender', 'InternetService', 'TechSupport', 'Contract']], drop_first=True)
print('Shape:', multi.shape)
multi.head()
```
