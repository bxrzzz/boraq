```
# Import required libraries, load the Telco dataset from URL,
# and convert 'TotalCharges' to numeric, filling missing values with 0.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce').fillna(0)
```

```
# Select 'MonthlyCharges' and 'TotalCharges' to observe their summary statistics
# and see the massive gap between their value ranges.

cols = ['MonthlyCharges', 'TotalCharges']
df[cols].describe().round(2)
```

```
# Initialize MinMaxScaler to scale the features into a fixed range of 0 to 1,
# then verify that the min and max values are exactly 0.0 and 1.0.

from sklearn.preprocessing import MinMaxScaler
mm = MinMaxScaler()
X = df[cols]
X_mm = mm.fit_transform(X)
X_mm_df = pd.DataFrame(X_mm, columns=cols)
X_mm_df.agg(['min', 'max']).round(2)
```

```
# Apply StandardScaler to center the data around a mean of 0
# and scale it to a standard deviation of 1.

from sklearn.preprocessing import StandardScaler
std = StandardScaler()
X_std = std.fit_transform(X)
X_std_df = pd.DataFrame(X_std, columns=cols)
X_std_df.describe().round(2)
```

```
# Use RobustScaler which scales the data using median and IQR (Interquartile Range)
# making it more resilient to outliers.

from sklearn.preprocessing import RobustScaler
rb = RobustScaler()
X_rb = rb.fit_transform(X)
X_rb_df = pd.DataFrame(X_rb, columns=cols)
X_rb_df.describe().round(2)
```

```
# Plot the original and scaled distributions side-by-side using histograms
# to confirm that the distribution shape remains identical while the x-axis scale changes.

fig, ax = plt.subplots(1, 4, figsize=(16, 4))
ax[0].hist(df['MonthlyCharges'], bins=30); ax[0].set_title('Original')
ax[1].hist(X_mm_df['MonthlyCharges'], bins=30); ax[1].set_title('MinMax')
ax[2].hist(X_std_df['MonthlyCharges'], bins=30); ax[2].set_title('Standard')
ax[3].hist(X_rb_df['MonthlyCharges'], bins=30); ax[3].set_title('Robust')
plt.tight_layout()
plt.show()
```

```
# To prevent data leakage, we split the data into train and test sets first.
# Then, we fit and transform on the train set, but only transform the test set.

from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=42)

scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

print('Train mean:', X_train_s.mean(axis=0).round(2))
print('Test mean:', X_test_s.mean(axis=0).round(2))
```

```
# Import pandas library and load the Telco churn dataset directly from the URL.

import pandas as pd
url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
```

```
# Filter and identify all columns with 'object' (text) data type,
# then print the total count and the column names.

obj_cols = df.select_dtypes(include='object').columns.tolist()
print('Total object columns:', len(obj_cols))
print(obj_cols)
```

```
# Check the frequency distribution of categories within the 'gender' column
# to see how the data is split between Male and Female.

df['gender'].value_counts()
```

```
# Perform one-hot encoding on 'InternetService' using a specific prefix
# to dynamically create binary columns for each category.

dummies = pd.get_dummies(df['InternetService'], prefix='Internet')
print('Shape:', dummies.shape)
dummies.head()
```

```
# Apply one-hot encoding on 'InternetService' again, but drop the first category
# to prevent multi-collinearity issues.

dummies2 = pd.get_dummies(df['InternetService'], prefix='Internet', drop_first=True)
print('With drop_first:', dummies2.shape)
dummies2.head()
```

```
# Use scikit-learn's OneHotEncoder to transform the 'TechSupport' column
# into a dense DataFrame with meaningful column labels.

from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output=False)
encoded = ohe.fit_transform(df[['TechSupport']])
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(['TechSupport']))
encoded_df.head()
```

```
# Map text values 'Yes' and 'No' into binary digits 1 and 0 using a dictionary mapping,
# then display the transformed column.

df['Partner_bin'] = df['Partner'].map({'No': 0, 'Yes': 1})
df[['Partner', 'Partner_bin']].head()
```

```
# Encode multiple categorical columns simultaneously using pd.get_dummies
# with drop_first set to True to optimize features for ML models.

multi = pd.get_dummies(df[['gender', 'InternetService', 'TechSupport', 'Contract']], drop_first=True)
print('Shape:', multi.shape)
multi.head()
```
