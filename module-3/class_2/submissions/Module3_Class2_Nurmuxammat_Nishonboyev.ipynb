```python
# This code prepares the dataset for analysis.
# First, it imports the required libraries for data analysis,
# numerical calculations, and data visualization.
# Then, it loads the customer churn dataset from a public URL.
# Next, it converts the TotalCharges column to numeric values.
# Any invalid or missing values are replaced with 0.
# Finally, it prints the dataset size (rows and columns)
# to confirm that the data was loaded successfully.

# === SETUP — run this first ===
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings('ignore')

url = "https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv"


df = pd.read_csv(url)

df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
).fillna(0)

print('Loaded:', df.shape)
```

```python
# This code selects three numerical columns from the dataset.
# The describe() function calculates summary statistics for these columns.
# It shows the count, mean, standard deviation, minimum,
# quartiles (25%, 50%, 75%), and maximum values.
# The round(2) function rounds all results to two decimal places
# to make the output easier to read.

cols = ['tenure', 'MonthlyCharges', 'TotalCharges']

df[cols].describe().round(2)
```

```python
# This code standardizes the selected numerical columns.
# First, it imports the StandardScaler class from scikit-learn.
# Then, it creates a scaler object.
# The selected numerical columns are stored in X.
# The scaler transforms the data so that each column has
# a mean of 0 and a standard deviation of 1.
# The standardized data is converted back into a DataFrame.
# Finally, summary statistics are displayed to verify
# that the scaling was successful.

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X = df[cols]

X_std = scaler.fit_transform(X)

X_std_df = pd.DataFrame(X_std, columns=cols)

X_std_df.describe().round(2)
```

```python
# This code applies Robust Scaling to the selected numerical columns.
# First, it imports the RobustScaler class from scikit-learn.
# Then, it creates a RobustScaler object.
# The selected numerical data is scaled using the median
# and the interquartile range (IQR).
# This method is less affected by outliers than StandardScaler.
# The scaled data is converted into a new DataFrame.
# Finally, it displays summary statistics to check the scaled values.

from sklearn.preprocessing import RobustScaler

rb = RobustScaler()

X_rb = rb.fit_transform(X)

X_rb_df = pd.DataFrame(X_rb, columns=cols)

X_rb_df.describe().round(2)
```

```python
# This code splits the dataset into training and testing sets.
# First, it imports the train_test_split function.
# Then, it divides the data into 80% training data
# and 20% testing data.
# Next, it creates a StandardScaler object.
# The scaler is fitted only on the training data
# and then used to transform both the training and testing sets.
# This prevents data leakage and ensures a fair evaluation.
# Finally, it prints the shapes of the training and testing data
# and checks that the training data has a mean close to 0.

from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(
    X,
    test_size=0.2,
    random_state=42
)

scaler = StandardScaler()

X_train_s = scaler.fit_transform(X_train)

X_test_s = scaler.transform(X_test)

print(f'Train shape: {X_train_s.shape}, Test shape: {X_test_s.shape}')
print(f'Train mean (should be ~0): {X_train_s.mean(axis=0).round(2)}')
```

```python
# This code saves the trained StandardScaler to a file.
# First, it imports the joblib library.
# Then, it saves the fitted scaler as 'my_scaler.joblib'.
# After that, it loads the saved scaler back into memory.
# Finally, it prints the mean values learned by the scaler
# to confirm that the scaler was saved and loaded correctly.

import joblib

joblib.dump(scaler, 'my_scaler.joblib')

loaded = joblib.load('my_scaler.joblib')

print('Saved and loaded. Learned mean:', loaded.mean_.round(2))
```
