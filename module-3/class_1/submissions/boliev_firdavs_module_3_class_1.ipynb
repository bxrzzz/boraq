## What this notebook covers
- Loaded the Telco Customer Churn dataset from a public CSV link.
- Checked missing values and duplicate rows before changing the data.
- Practiced safe filling with median values and zero values where appropriate.
- Converted and cleaned columns so the dataset is easier to analyze later.


```python
# This setup cell imports the libraries, loads the Telco dataset, and converts TotalCharges into a numeric column.
# === SETUP — run this first ===
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings; warnings.filterwarnings('ignore')

url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce').fillna(0)
print('Loaded:', df.shape)
```

### Cell 1 — see how many missing values per column

```python
# This cell counts missing values in every column so we can see which columns need cleaning.
df.isnull().sum()
```

### Cell 2 — check how many duplicate rows we have

```python
# This cell checks whether the dataset has any fully duplicated rows.
print('Duplicates:', df.duplicated().sum())
```

### Cell 3 — make a copy and add a fake missing value to test fillna

```python
# This cell creates a copy of the data and adds fake missing tenure values so we can practice fillna safely.
df_test = df.copy()
df_test.loc[0:9, 'tenure'] = np.nan
print('Missing tenure now:', df_test['tenure'].isnull().sum())
```

### Cell 4 — fill missing tenure with the median

```python
# This cell fills the missing tenure values with the median because the median is stable when outliers exist.
median_tenure = df_test['tenure'].median()
df_test['tenure'] = df_test['tenure'].fillna(median_tenure)
print('After fillna, missing:', df_test['tenure'].isnull().sum())
print('Median used:', median_tenure)
```

### Cell 5 — fill missing TotalCharges with 0

```python
# This cell fills missing TotalCharges values with 0 so the column has no empty numeric values.
df['TotalCharges'] = df['TotalCharges'].fillna(0)
print('TotalCharges missing after fillna:', df['TotalCharges'].isnull().sum())
```

### Cell 6 — drop duplicate rows

```python
# This cell removes duplicate rows and prints how many rows were dropped.
before = len(df)
df = df.drop_duplicates()
after = len(df)
print(f'Before: {before}, After: {after}, Dropped: {before - after}')
```

### Cell 7 — convert SeniorCitizen from 0/1 to 'No'/'Yes'

```python
# This cell changes SeniorCitizen from 0 and 1 into readable No and Yes labels.
df['SeniorCitizen'] = df['SeniorCitizen'].map({0: 'No', 1: 'Yes'})
df['SeniorCitizen'].value_counts()
```

### Cell 8 — strip whitespace from gender column

```python
# This cell removes extra spaces from gender values so categories are clean and consistent.
df['gender'] = df['gender'].str.strip()
df['gender'].value_counts()
```
