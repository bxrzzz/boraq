```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

The missingness in TotalCharges is classified as MAR because the probability of the value being missing is systematically related to observed data in another column—specifically, the tenure column

```python
df['TotalCharges_missing'] = df['TotalCharges'].isnull().astype(int)
df.fillna({"TotalCharges": df['TotalCharges'].median()}, inplace=True)
cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    if df[col].isnull().any():
        df[col].fillna(df[col].mode()[0], inplace=True)
```
