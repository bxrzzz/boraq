```python
# Step 0 --- LOad the Titanic dataset
# This code imports the classic Titanic dataset from a public remote repository into a Pandas DataFrame.
# It then outputs the overall matrix dimensions of the dataset and prints a visual preview of its first 5 rows.

import pandas as pd
import numpy as np
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)
print('Shape:', df.shape)
df.head()
```

```python
# Step 1 --- Find missing values
# Check for missing data across the entire DataFrame

df.isnull().sum()
```

```python
# Step 2 --- Fill missing Age with the median
# This code calculates the median age of passengers and uses it to replace any missing values in the 'Age' column.
# It then prints the median value applied and verifies that the total count of null values in that column is successfully reduced to zero.

median_age = df['Age'].median()
df['Age'] = df['Age'].fillna(median_age)
print('Median used:', median_age)
print('Missing Age now:', df['Age'].isnull().sum())
```

```python
# Step 3 --- Fill missing Embarked with the mode(most common value)
# mode() returns a Series; use [0] to get the first value
# # This code finds the most frequent departure port (mode) and uses it to fill in any missing entries in the 'Embarked' column.
# It then displays the chosen mode and confirms that there are no remaining missing values left in that column.

mode_embarkked = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(mode_embarkked)
print('Mode used:', mode_embarkked)
print('Missing Embarked:', df['Embarked'].isnull().sum())
```

```python
# Step 4 --- Drop the Cabin column (too much missing)
# This code drops the 'Cabin' column from the DataFrame because it contains too many missing values to be useful.
# It then prints the updated number of columns remaining in the dataset to confirm the successful deletion.

df = df.drop(columns=['Cabin'])
print('Columns now:', df.shape[1])
```

```python
# Step 5 --- Check for duplicate rows
# duplicated() returns True/False per row; .sum() counts the Trues
# It returns the total count of duplicated rows present in the DataFrame to help ensure data integrity.

print('Duplicates:', df.duplicated().sum())
```

```python
# Step 6 --- Convert Survived from 0/1 to 'No'/'Yes'
# Map binary survival integers to descriptive text labels for better readability

df['Survived_label'] = df['Survived'].map({0: 'No', 1:'Yes'})
df[['Survived', 'Survived_label']].head()
```

```python
# Step 7 --- Strip whitespace from Sex column
# Use str.strip() to remove leading/trailing spaces.
# Clean the text format of the Sex column and check the value distribution

df['Sex'] = df['Sex'].str.strip().str.lower()
df['Sex'].value_counts()
```

```python
# Step 8 --- Final check --- confirmdataset is clean
# This code prints a comprehensive summary of any remaining missing values across all columns in the dataset.
# It then displays the final matrix dimensions of the cleaned DataFrame to confirm the data shape before modeling.

print('Missing per column:')
print(df.isnull().sum())
print()
print('Final shape:', df.shape)
```
