````python
# === SETUP — run this first ===
import pandas as pd         #uploading python libraries
import numpy as np
import matplotlib.pyplot as plt
import warnings; warnings.filterwarnings('ignore')

url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce').fillna(0)
print('Loaded:', df.shape)
````

### Cell 1 — see how many missing values per column

````python
df.isnull().sum()  #it is for seeing how many missing values in each column
````

### Cell 2 — check how many duplicate rows we have

````python
print('Duplicates:', df.duplicated().sum())  #it shows how many duplicated values exist
````

### Cell 3 — make a copy and add a fake missing value to test fillna

````python
# Create a test dataframe to simulate missing data for testing imputation techniques
df_test = df.copy()

# Introduce 10 missing values (NaN) in the 'tenure' column (rows 0 to 9)
df_test.loc[0:9, 'tenure'] = np.nan

# Verify that exactly 10 missing values were successfully introduced
print('Missing tenure now:', df_test['tenure'].isnull().sum())
````

### Cell 4 — fill missing tenure with the median

````python
median_tenure = df_test['tenure'].median()          #filling those 10 fake missing values with the median of tenure column values
df_test['tenure'] = df_test['tenure'].fillna(median_tenure)
print('After fillna, missing:', df_test['tenure'].isnull().sum())   #seeing if it has more missing values here
print('Median used:', median_tenure)             #these are being done for testing the model
````

### Cell 5 — fill missing TotalCharges with 0

````python
df['TotalCharges'] = df['TotalCharges'].fillna(0)  # Verify that all missing values in TotalCharges have been successfully filled with 0
print('TotalCharges missing after fillna:', df['TotalCharges'].isnull().sum())
````

### Cell 6 — drop duplicate rows

````python
before = len(df) #it count the rows in the data
df = df.drop_duplicates() #it drops duplicated rows
after = len(df) #again, it is counting rows after dropping
print(f'Before: {before}, After: {after}, Dropped: {before - after}')
````

### Cell 7 — convert SeniorCitizen from 0/1 to 'No'/'Yes'

````python
df['SeniorCitizen'] = df['SeniorCitizen'].map({0: 'No', 1: 'Yes'})#it is converting 1 and 0 values to Yes and No
df['SeniorCitizen'].value_counts()  #then , it is counting how many times each value appears in data
````

### Cell 8 — strip whitespace from gender column

````python
df['gender'] = df['gender'].str.strip() #it removes any extra space in the values written
df['gender'].value_counts()  #then, it learns how many times each value in that column appears
````
