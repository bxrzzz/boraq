```python
# This code prepares the dataset for analysis.
# First, it imports the required libraries for data processing and visualization.
# Then, it loads the customer churn dataset from a public URL.
# The TotalCharges column is converted to numeric values.
# Any missing or invalid values are replaced with 0 to avoid errors.
# Finally, the code prints the number of rows and columns to confirm
# that the dataset was loaded successfully.

# === SETUP — run this first ===
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'

df = pd.read_csv(url)

df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
).fillna(0)

print('Loaded:', df.shape)
```

```python
# This code checks for missing values in the dataset.
# It counts the number of missing values in each column.
# This helps us identify columns that may need data cleaning.
# If a column shows 0, it has no missing values.
# If the number is greater than 0, the missing values should be handled.

df.isnull().sum()
```

```python
# This code checks for duplicate rows in the dataset.
# It counts how many rows are exactly the same.
# Duplicate rows can affect the accuracy of data analysis.
# The result is printed to show the total number of duplicates.
# If the result is 0, the dataset has no duplicate rows.

print('Duplicates:', df.duplicated().sum())
```

```python
# This code creates a copy of the original dataset for testing.
# It sets the first 10 values in the 'tenure' column to missing (NaN).
# This allows us to practice handling missing values
# without changing the original dataset.
# Finally, it counts and prints the number of missing values
# in the 'tenure' column.

df_test = df.copy()

df_test.loc[0:9, 'tenure'] = np.nan

print('Missing tenure now:', df_test['tenure'].isnull().sum())
```

```python
# This code fills the missing values in the 'tenure' column.
# First, it calculates the median value of the tenure column.
# Then, it replaces all missing values (NaN) with the median.
# This keeps the dataset complete without changing the original distribution too much.
# Finally, it prints the number of remaining missing values
# and the median value that was used to fill them.

median_tenure = df_test['tenure'].median()

df_test['tenure'] = df_test['tenure'].fillna(median_tenure)

print('After fillna, missing:', df_test['tenure'].isnull().sum())
print('Median used:', median_tenure)
```

```python
# This code fills any missing values in the TotalCharges column with 0.
# First, it checks for missing values in the column.
# Then, it replaces all missing values with 0 using fillna().
# Finally, it prints the number of missing values remaining.
# If the result is 0, the column has no missing values.

df['TotalCharges'] = df['TotalCharges'].fillna(0)

print('TotalCharges missing after fillna:', df['TotalCharges'].isnull().sum())
```

```python
# This code removes duplicate rows from the dataset.
# First, it counts the number of rows before removing duplicates.
# Then, it removes all duplicate rows using drop_duplicates().
# After that, it counts the number of rows again.
# Finally, it prints the number of rows before cleaning,
# the number of rows after cleaning, and how many duplicate rows were removed.

before = len(df)

df = df.drop_duplicates()

after = len(df)

print(f'Before: {before}, After: {after}, Dropped: {before - after}')
```

```python
# This code changes the values in the SeniorCitizen column.
# First, it replaces 0 with "No" and 1 with "Yes".
# This makes the data easier to read and understand.
# Then, it counts how many customers are in each group.
# Finally, it displays the number of senior and non-senior customers.

df['SeniorCitizen'] = df['SeniorCitizen'].map({0: 'No', 1: 'Yes'})

df['SeniorCitizen'].value_counts()
```

```python
# This code cleans the values in the gender column.
# First, it removes any extra spaces before or after each value.
# This helps keep the data consistent and avoids incorrect categories.
# Then, it counts how many customers belong to each gender.
# Finally, it displays the number of customers in each group.

df['gender'] = df['gender'].str.strip()

df['gender'].value_counts()
```
