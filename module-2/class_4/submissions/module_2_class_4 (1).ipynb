```python
# CSV faylni internetdan yuklab, TotalCharges ustunidagi matn qiymatlarni raqamga aylantiryapman
import pandas as pd
url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce').fillna(0)
print('Shape:', df.shape)
```

```python
# Har bir ustunning nomi, ma'lumot turi va bo'sh bo'lmagan qiymatlar sonini ko'ryapman
df.info()
```

```python
# Raqamli ustunlar uchun statistika (min, max, o'rtacha va h.k.) ni 2 xonagacha yaxlitlab chiqaryapman
df.describe().round(2)
```

```python
# Har bir ustunda nechta noyob (takrorlanmagan) qiymat borligini hisoblayapman
df.nunique()
```

```python
# Oylik to'lovi 80 dollardan katta bo'lgan mijozlarni filtrlab, sonini va birinchi 5 tasini ko'rsataman
high_pay = df[df['MonthlyCharges'] > 80]
print('High-paying customers:', len(high_pay))
high_pay.head()
```

```python
# Mijozlarni tenure (necha oy xizmatda bo'lgani) bo'yicha kamayish tartibida saralab, eng yuqori 10 tasini olayapman
top10 = df.sort_values('tenure', ascending=False).head(10)
top10[['customerID', 'tenure', 'Contract', 'MonthlyCharges']]
```

```python
# To'lov usuli bo'yicha guruhlab, har bir guruhning o'rtacha oylik to'lovini hisoblayapman
df.groupby('PaymentMethod')['MonthlyCharges'].mean().round(2)
```

```python
# Shartnoma turi va mijozning ketish holati (Churn) o'rtasidagi bog'liqlikni jadval qilib ko'rsataman
pd.crosstab(df['Contract'], df['Churn'])
```
