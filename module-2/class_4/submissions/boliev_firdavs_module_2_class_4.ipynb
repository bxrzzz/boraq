This notebook practices DataFrame inspection, filtering, sorting, grouping, and crosstab analysis with the Telco customer churn dataset.

## Step 0: Load the Data

```python
# Load the Telco dataset and make TotalCharges numeric for analysis.
import pandas as pd

url = "https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv"
df = pd.read_csv(url)
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce").fillna(0)

print("Shape:", df.shape)
df.head()
```

## Step 1: See Column Info

```python
# Show column names, non-null counts, and data types.
df.info()
```

## Step 2: Statistics for Numeric Columns

```python
# Describe numeric columns so we can see count, mean, min, and max values.
df.describe().round(2)
```

## Step 3: Count Unique Values per Column

```python
# Count distinct values in every column.
df.nunique()
```

## Step 4: Filter Customers Paying More Than $80 per Month

```python
# Keep customers whose MonthlyCharges value is greater than 80.
high_pay = df[df["MonthlyCharges"] > 80]

print("High-paying customers:", len(high_pay))
high_pay.head()
```

## Step 5: Sort Top 10 by Tenure

```python
# Sort customers by tenure from highest to lowest and show the first 10 rows.
top10 = df.sort_values("tenure", ascending=False).head(10)
top10[["customerID", "tenure", "Contract", "MonthlyCharges"]]
```

## Step 6: Average Monthly Charges per Payment Method

```python
# Group customers by PaymentMethod and calculate average MonthlyCharges.
df.groupby("PaymentMethod")["MonthlyCharges"].mean().round(2)
```

## Step 7: TotalCharges Mean and Count by Churn

```python
# Group by Churn and calculate both mean and count for TotalCharges.
df.groupby("Churn").agg({"TotalCharges": ["mean", "count"]}).round(2)
```

## Step 8: Crosstab of Contract vs Churn

```python
# Count how many customers churned or did not churn for each Contract type.
pd.crosstab(df["Contract"], df["Churn"])
```
