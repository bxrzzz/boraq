```
# Import pandas library and load the Telco Customer Churn dataset from GitHub.
# We convert 'TotalCharges' to numeric because it contains some non-numeric values,
# and fill any missing values with 0 to avoid errors in analysis.
import pandas as pd

url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce').fillna(0)
print('Shape:', df.shape)
# Expected output: Shape: (7043, 21)
```
