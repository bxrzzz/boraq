```python
import pandas as pd
url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url) #it opens the file
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce').fillna(0)  #it is turning text into number, missing values in to NaN, Then it is turning this NaN values into 0   #uploading python libraries to work with
print('Shape:', df.shape) #it shows how many rows and columns we have
```

```python
df.info() #it shows what kind of datatype each column has
```

```python
df.describe().round(2) # Calculate key summary statistics for numerical columns and round the results to 2 decimal places
```

```python
df.nunique() #it showshow many different values each column has
```

```python
high_pay = df[df['MonthlyCharges'] > 80]  #it is creating the column called high_pay which only includes the values above 80 from monthly charges column
print('High-paying customers:', len(high_pay)) #it count how many these high pay custumers there are
high_pay.head() #it shows the first 5 rows
```

```python
top10 = df.sort_values('tenure', ascending=False).head(10) #it shows 10 tenure values from highest to the lowest
top10[['customerID', 'tenure', 'Contract', 'MonthlyCharges']] # Create a cleaner view of the 'top10' table by selecting and displaying only these 4 specific columns
```

```python
df.groupby('PaymentMethod')['MonthlyCharges'].mean().round(2)  #it calculates the average value of monthly charges for each payment method
```

```python
df.groupby('Churn').agg({'TotalCharges': ['mean', 'count']}).round(2) #it calculates the average value of total charges for churn, rounded to 2 decimal numbers
```

```python
pd.crosstab(df['Contract'], df['Churn']) #Cross-tabulation to count the number of stayed (No) and churned (Yes) customers for each contract type
```
