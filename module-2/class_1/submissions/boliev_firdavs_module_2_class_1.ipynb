## Task 3: Function Practice

The `calculate_average` function receives a list of numbers and returns their average. Reusing a function is better than writing the same formula again and again.

```python
# This function calculates the average of any non-empty list of numbers.
def calculate_average(numbers):
    total = sum(numbers)
    count = len(numbers)
    return total / count

test1 = [80, 90, 70, 85, 95]
test2 = [55, 62, 48, 71]

print("Average 1:", calculate_average(test1))
print("Average 2:", calculate_average(test2))
```

## Task 4: Loop with `enumerate()`

`enumerate()` is useful when we need both the position of an item and the item value itself.

```python
# This loop prints each city with its index number.
cities = ["Tashkent", "Samarkand", "Bukhara", "Namangan", "Fergana"]

for index, city in enumerate(cities):
    print(f"Index {index}: {city}")
```

## Task 5: Combine Everything

This final task creates exam scores, calculates the class average, and checks whether the class passed using a threshold of 60.

```python
# These made-up exam scores make the result reproducible every time the cell runs.
exam_scores = [72, 88, 91, 64, 55, 79, 83, 69, 94, 76]
average_score = calculate_average(exam_scores)

print("Scores:", exam_scores)
print(f"Class average: {average_score:.1f}")

if average_score >= 60:
    print("Result: PASS")
else:
    print("Result: FAIL")
```
