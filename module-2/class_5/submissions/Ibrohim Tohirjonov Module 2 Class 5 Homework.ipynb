````python
# === SETUP — run this first ===
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings; warnings.filterwarnings('ignore')

url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce').fillna(0)
print('Loaded:', df.shape)

# This script loads the Telco Customer Churn dataset from a public IBM repository into a Pandas DataFrame.
# It also ensures the 'TotalCharges' column is converted to a numeric format, replacing any missing values with zero.
````

### Cell 1 — bar chart of Churn count

````python
# Calculate counts and set up the visualization
counts = df['Churn'].value_counts()
plt.figure(figsize=(6, 4))

# Plot the distribution as a bar chart with custom colors
plt.bar(counts.index, counts.values, color=['#2E86AB', '#E63946'])
plt.title('Churned vs Stayed')
plt.ylabel('Number of customers')
plt.show()

# This code computes the total counts of churned versus retained customers from the 'Churn' column.
# It then visualizes the distribution using a customized bar chart to easily compare the two customer groups.
````

### Cell 2 — histogram of MonthlyCharges

````python
# This code generates a histogram to visualize the overall distribution of customers' monthly billing amounts.
# It divides the 'MonthlyCharges' values into 30 equal-width bins with custom colors and edges to show where customer spending peaks.

plt.figure(figsize=(8, 4))
plt.hist(df['MonthlyCharges'], bins=30, color='#2E86AB', edgecolor='black')
plt.title('Distribution of MonthlyCharges')
plt.xlabel('Monthly Charges ($)')
plt.ylabel('Number of customers')
plt.show()
````

### Cell 3 — scatter of tenure vs TotalCharges

````python
# This code creates a scatter plot to analyze the relationship between customer tenure and their total charges over time.
# It uses transparency (alpha=0.3) and small marker sizes (s=10) to clearly handle high data density and reveal overlapping patterns.

plt.figure(figsize=(8, 5))
plt.scatter(df['tenure'], df['TotalCharges'], alpha=0.3, s=10)
plt.title('Tenure vs Total Charges')
plt.xlabel('Tenure (months)')
plt.ylabel('Total Charges ($)')
plt.show()
````

### Cell 4 — box plot of tenure by Churn

````python
# This code splits customer tenure data into two separate groups based on whether they stayed or churned.
# It then generates side-by-side boxplots to compare the range, median, and overall distribution of loyalty months between both groups.

churn_yes = df[df['Churn'] == 'Yes']['tenure']
churn_no = df[df['Churn'] == 'No']['tenure']
plt.figure(figsize=(6, 5))
plt.boxplot([churn_no, churn_yes], labels=['Stayed', 'Churned'])
plt.title('Tenure distribution by Churn')
plt.ylabel('Tenure (months)')
plt.show()
````

### Cell 5 — 2x2 subplots dashboard

````python
# Initialize a 2x2 dashboard layout and plot all univariate/bivariate features together
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

# Top-Left: Churn bar chart
ax[0, 0].bar(df['Churn'].value_counts().index, df['Churn'].value_counts().values, color='steelblue')
ax[0, 0].set_title('Churn')

# Top-Right: Monthly charges distribution histogram
ax[0, 1].hist(df['MonthlyCharges'], bins=30, color='orange')
ax[0, 1].set_title('MonthlyCharges')

# Bottom-Left: Tenure vs Monthly Charges relationship scatter plot
ax[1, 0].scatter(df['tenure'], df['MonthlyCharges'], alpha=0.2, s=5)
ax[1, 0].set_title('Tenure vs MonthlyCharges')

# Bottom-Right: Tenure distribution histogram
ax[1, 1].hist(df['tenure'], bins=30, color='green')
ax[1, 1].set_title('Tenure')

# Optimize spacing and display the dashboard
plt.tight_layout()
plt.show()

# This code initializes a 2x2 grid layout using Matplotlib subplots to construct multiple data visualizations at once.
# It populates the grid with a bar chart, two separate histograms, and a scatter plot to analyze customer churn, spending, and tenure patterns in a single dashboard view.
````
