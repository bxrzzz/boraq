```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```
data = {
    "sales": [100, 200, 150, 300, 400, 250, 500, 600, 700, 800],
    "profit": [20, 50, 30, 80, 120, 60, 150, 200, 250, 300],
    "category": ["A", "B", "A", "B", "A", "B", "A", "B", "A", "B"],
    "region": ["East", "West", "East", "West", "East", "West", "East", "West", "East", "West"]
}

df = pd.DataFrame(data)
df.head()
```

## Section 2

```
# Histogram
plt.hist(df["sales"], bins=5)
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Sales Distribution")
plt.show()
```

```
# Scatter
plt.scatter(df["sales"], df["profit"])
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Sales vs Profit")
plt.show()
```

```
# Seaborn plots
sns.histplot(df, x="sales", kde=True)
plt.title("Sales Histogram")
plt.show()

sns.boxplot(data=df, x="category", y="sales")
plt.title("Sales by Category")
plt.show()

sns.scatterplot(data=df, x="sales", y="profit", hue="region")
plt.title("Sales vs Profit by Region")
plt.show()

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```
# OO Matplotlib
x = df["sales"]
y = df["profit"]

fig, ax = plt.subplots(figsize=(8,5))
ax.plot(x, y)

ax.set_xlabel("Sales")
ax.set_ylabel("Profit")
ax.set_title("Line Plot")

fig.savefig("out.png", dpi=150)
plt.show()
```
