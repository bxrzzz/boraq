# Module 5 | Class 2 Assignment: When K-Means Fails

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
```

## Task 1: Generate and Plot the Data

```python
X, y = make_moons(n_samples=300, noise=0.1, random_state=42)

plt.figure(figsize=(6, 4))
plt.scatter(X[:, 0], X[:, 1])
plt.title("Task 1: Moon-shaped Data")
plt.tight_layout()
plt.show()
```

## Task 2: Apply K-Means (k=2)

```python
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(X)

plt.figure(figsize=(6, 4))
plt.scatter(X[:, 0], X[:, 1], c=kmeans.labels_, cmap='viridis')
plt.title("Task 2: K-Means (k=2)")
plt.tight_layout()
plt.show()
```

## Task 3: Apply DBSCAN

```python
dbscan = DBSCAN(eps=0.2, min_samples=5)
dbscan.fit(X)

plt.figure(figsize=(6, 4))
plt.scatter(X[:, 0], X[:, 1], c=dbscan.labels_, cmap='viridis')
plt.title("Task 3: DBSCAN (eps=0.2, min_samples=5)")
plt.tight_layout()
plt.show()
```

## Task 4: Explain the Difference

K-Means defines a cluster as a group of points closest to a centroid, which means it assumes clusters are spherical (globular) in shape. Because the moon-shaped data forms two crescent curves rather than two round blobs, K-Means cannot correctly separate them — it simply splits the data by proximity to two center points, cutting across both crescents.

DBSCAN, on the other hand, defines a cluster as a dense region of points connected to each other through neighborhoods (eps-radius). It does not assume any particular shape, so it can follow the curved density of each crescent and correctly identify them as two separate clusters. This makes DBSCAN far better suited for data with arbitrary, non-spherical shapes.

## Task 5 (Bonus): Agglomerative Clustering

```python
agg = AgglomerativeClustering(n_clusters=2)
agg.fit(X)

plt.figure(figsize=(6, 4))
plt.scatter(X[:, 0], X[:, 1], c=agg.labels_, cmap='viridis')
plt.title("Task 5 (Bonus): Agglomerative Clustering (n_clusters=2)")
plt.tight_layout()
plt.show()
```
