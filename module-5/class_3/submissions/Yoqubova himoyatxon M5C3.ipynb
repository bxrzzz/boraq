# Module 5 | Class 3 Assignment: Dimensionality Reduction with PCA

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
```

## Load Dataset

```python
# Download from: https://www.kaggle.com/datasets/vjchoudhary7/customer-segmentation-tutorial-in-python
df = pd.read_csv('Mall_Customers.csv')
print(df.head())
print(df.shape)
```

## Prepare Features and Cluster Labels (from Activity 1)

```python
# Select numeric features
feature_names = ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']
X = df[feature_names].values

# Assign cluster labels using K-Means (from Class 1 Activity 1)
kmeans = KMeans(n_clusters=5, random_state=42)
cluster_labels = kmeans.fit_predict(X)
```

## Task 1: Fit PCA and Plot Explained Variance

```python
# Step 1: Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 2: Fit PCA with all components
pca_full = PCA(n_components=None)
pca_full.fit(X_scaled)

# Step 3: Plot scree plot (explained variance ratio)
plt.figure(figsize=(6, 4))
plt.bar(
    range(1, len(pca_full.explained_variance_ratio_) + 1),
    pca_full.explained_variance_ratio_
)
plt.xlabel('Principal Component')
plt.ylabel('Explained Variance Ratio')
plt.title('Task 1: Scree Plot — Explained Variance per Component')
plt.tight_layout()
plt.show()

# Step 4: Cumulative variance
cumulative = np.cumsum(pca_full.explained_variance_ratio_)
print("Cumulative explained variance:", cumulative)
n_90 = np.argmax(cumulative >= 0.90) + 1
n_95 = np.argmax(cumulative >= 0.95) + 1
print(f"Components needed for 90% variance: {n_90}")
print(f"Components needed for 95% variance: {n_95}")
```

**Answer:** Based on the cumulative explained variance printed above, we can see how many principal components are needed to capture 90% and 95% of the total variance in the Mall Customers dataset.

## Task 2: Reduce to 2D and Visualize

```python
# Step 1: Fit PCA with 2 components and transform
pca_2d = PCA(n_components=2, random_state=42)
X_pca = pca_2d.fit_transform(X_scaled)

# Step 2: Scatter plot of PC1 vs PC2, colored by cluster labels
plt.figure(figsize=(7, 5))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=cluster_labels, cmap='viridis', alpha=0.8)
plt.colorbar(scatter, label='Cluster')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('Task 2: PCA 2D Projection Colored by Cluster Labels')
plt.tight_layout()
plt.show()
```

**Observation:** The scatter plot above shows the data projected onto PC1 and PC2. Since the Mall Customers dataset has only 3 features and relatively simple structure, some clusters may separate reasonably well in PCA space. However, because PCA maximizes variance rather than cluster separability, the projection may not perfectly align with the K-Means cluster boundaries found in the original feature space.

## Task 3: Interpret the Principal Components

```python
# Step 1: Print PCA loadings as a DataFrame
loadings = pd.DataFrame(
    pca_2d.components_,
    columns=feature_names,
    index=['PC1', 'PC2']
)
print(loadings)
```

**Interpretation:**

- **PC1** captures the direction of greatest variance in the data. By looking at the loadings above, the features with the largest absolute values contribute most to PC1. For example, if `Annual Income (k$)` and `Spending Score (1-100)` have high loadings, PC1 could be interpreted as **"purchasing power"** — a combined measure of how much customers earn and spend.

- **PC2** captures the second direction of variance, orthogonal to PC1. If `Age` has a high loading on PC2, it could be interpreted as **"customer life stage"** — distinguishing younger from older customers independent of their income or spending habits.
