```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

## Task 1: Plot the Data Points

**What to do:** Plot these 6 points on a 2D grid: `(1,2), (2,1), (1.5,1.5), (8,8), (9,7), (8.5,8.5)`

**Steps:**
- Step 1: Draw a 10x10 grid
- Step 2: Mark each point with a dot and label it
- Step 3: Visually identify which points look like they belong together

```python
points = np.array([[1,2],[2,1],[1.5,1.5],[8,8],[9,7],[8.5,8.5]], dtype=float)
labels_str = ['(1,2)','(2,1)','(1.5,1.5)','(8,8)','(9,7)','(8.5,8.5)']

fig, ax = plt.subplots(figsize=(6,6))
ax.scatter(points[:,0], points[:,1], s=100, color='steelblue', zorder=3)
for i, (x, y) in enumerate(points):
    ax.annotate(labels_str[i], (x, y), textcoords='offset points', xytext=(8,4), fontsize=9)
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_xticks(range(11))
ax.set_yticks(range(11))
ax.grid(True, linestyle='--', alpha=0.5)
ax.set_title('Task 1: 6 Data Points on 2D Grid')
ax.set_xlabel('X')
ax.set_ylabel('Y')
plt.tight_layout()
plt.show()
```

## Task 2: Run K-Means Manually (K=2, 2 Iterations)

**What to do:** Execute K-Means by hand with starting centroids **(1,2)** and **(9,7)**

**Formula:** `distance = sqrt((x2-x1)^2 + (y2-y1)^2)`  — keep 2 decimal places

---

### Iteration 1

**Initial centroids:** C1 = (1, 2)  |  C2 = (9, 7)

**Step 1 — Distance calculations:**

| Point | dist to C1 (1,2) | dist to C2 (9,7) | Assigned Cluster |
|-------|-----------------|-----------------|------------------|
| (1, 2) | 0.00 | 9.43 | Cluster 1 |
| (2, 1) | 1.41 | 9.22 | Cluster 1 |
| (1.5, 1.5) | 0.71 | 9.30 | Cluster 1 |
| (8, 8) | 9.22 | 1.41 | Cluster 2 |
| (9, 7) | 9.43 | 0.00 | Cluster 2 |
| (8.5, 8.5) | 9.92 | 1.58 | Cluster 2 |

**Step 2 — Assignments:**
- Cluster 1: (1,2), (2,1), (1.5,1.5)
- Cluster 2: (8,8), (9,7), (8.5,8.5)

**Step 3 — New centroids:**
- New C1 = ((1+2+1.5)/3, (2+1+1.5)/3) = **(1.5, 1.5)**
- New C2 = ((8+9+8.5)/3, (8+7+8.5)/3) = **(8.5, 7.833)**

---

### Iteration 2

**Updated centroids:** C1 = (1.5, 1.5)  |  C2 = (8.5, 7.833)

**Step 1 — Distance calculations:**

| Point | dist to C1 (1.5,1.5) | dist to C2 (8.5,7.83) | Assigned Cluster |
|-------|---------------------|----------------------|------------------|
| (1, 2) | 0.71 | 9.50 | Cluster 1 |
| (2, 1) | 0.71 | 9.43 | Cluster 1 |
| (1.5, 1.5) | 0.00 | 9.44 | Cluster 1 |
| (8, 8) | 9.19 | 0.53 | Cluster 2 |
| (9, 7) | 9.30 | 0.97 | Cluster 2 |
| (8.5, 8.5) | 9.90 | 0.67 | Cluster 2 |

**Step 2 — Assignments remain the same:**
- Cluster 1: (1,2), (2,1), (1.5,1.5)
- Cluster 2: (8,8), (9,7), (8.5,8.5)

**Step 3 — New centroids:**
- New C1 = (1.5, 1.5) ← same as Iteration 1
- New C2 = (8.5, 7.833) ← same as Iteration 1

> ✅ **Centroids did not change — the algorithm has CONVERGED after 2 iterations.**

## Task 3: Verify with sklearn

**What to do:** Run KMeans in code with the same initial centroids to check manual work.

```python
points = np.array([[1,2],[2,1],[1.5,1.5],[8,8],[9,7],[8.5,8.5]], dtype=float)

kmeans = KMeans(n_clusters=2, init=np.array([[1,2],[9,7]]), n_init=1, random_state=42)
kmeans.fit(points)

print('Cluster labels :', kmeans.labels_)
print('Final centroids:', kmeans.cluster_centers_)
```

**Comparison:**

| | Manual Result | sklearn Result |
|-|--------------|----------------|
| Cluster 1 points | (1,2), (2,1), (1.5,1.5) | labels = 0 |
| Cluster 2 points | (8,8), (9,7), (8.5,8.5) | labels = 1 |
| Centroid 1 | (1.5, 1.5) | [1.5, 1.5] |
| Centroid 2 | (8.5, 7.833) | [8.5, 7.833] |

✅ Manual result matches sklearn output exactly.

## Task 4: Visualize the Clusters

**What to do:** Plot the final clusters with different colors and mark the centroids.

```python
labels = kmeans.labels_
centroids = kmeans.cluster_centers_
labels_str = ['(1,2)','(2,1)','(1.5,1.5)','(8,8)','(9,7)','(8.5,8.5)']

fig, ax = plt.subplots(figsize=(7, 7))

# Plot points colored by cluster
scatter = ax.scatter(points[:,0], points[:,1], c=labels,
                     cmap='bwr', s=120, zorder=3, edgecolors='black', linewidths=0.5)
for i, (x, y) in enumerate(points):
    ax.annotate(labels_str[i], (x, y), textcoords='offset points', xytext=(8, 4), fontsize=9)

# Plot centroids
ax.scatter(centroids[:,0], centroids[:,1], marker='X', s=250,
           c=['red','blue'], edgecolors='black', linewidths=1.5, zorder=5, label='Centroids')

from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='red',  label='Cluster 0'),
    Patch(facecolor='blue', label='Cluster 1'),
    plt.Line2D([0],[0], marker='X', color='w', markerfacecolor='gray',
               markersize=12, label='Centroids')
]
ax.legend(handles=legend_elements, loc='upper left')

ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_xticks(range(11))
ax.set_yticks(range(11))
ax.grid(True, linestyle='--', alpha=0.5)
ax.set_title('Task 4: Final Clusters (sklearn K-Means, K=2)')
ax.set_xlabel('X')
ax.set_ylabel('Y')
plt.tight_layout()
plt.show()
```
